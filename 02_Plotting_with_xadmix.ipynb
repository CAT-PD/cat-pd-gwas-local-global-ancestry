````python
Q <- read.csv("Q_and_sample_information.csv", header=TRUE, sep=",", stringsAsFactors=FALSE)
Q_subset <- Q[, c("IID", "country", paste0("POP", 1:10))]

palette_colors <- c('#FF97FF', '#00CC96', '#19D3F3', '#FFA15A', '#FECB52', '#B6E880', '#FF6692', '#636EFA', '#AB63FA', '#FECB52')


# Map abbreviated country codes
abbr_map <- c(
  "Kazakhstan" = "KAZ",
  "Tajikistan" = "TJK",
  "Armenia" = "ARM",
  "Azerbaijan" = "AZE",
  "Georgia" = "GEO"
)
````

````python
Q_subset$country <- factor(Q_subset$country, levels = names(abbr_map))
Q_subset$country <- as.character(Q_subset$country)
Q_subset$country[Q$country %in% names(abbr_map)] <- abbr_map[Q_subset$country[Q_subset$country %in% names(abbr_map)]]

Q_subset$country <- factor(Q_subset$country)
````

````python
p <- admix_barplot(Q_subset,
       K = 3:ncol(Q_subset),
       grouping = "country",
       sortkey = "POP1", # Sort by popmax
       names=FALSE,
       palette = palette_colors,
    )
````

````python
p <- p +
  theme(
    text = element_text(family = "Times New Roman", size = 28),
    legend.position = "none"
  )
````

````python
# Change width and height for accurate length
ggsave("admix_barplot_custom_palette_by_country.png", p, width = 48, height = 12, dpi = 300)
````
